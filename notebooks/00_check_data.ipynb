```python
import sys, os
sys.path.append(os.path.abspath(".."))  

from src.datasets_kaggle import get_kaggle_celeba_loaders, ATTRS
import matplotlib.pyplot as plt

train_loader, _ = get_kaggle_celeba_loaders(data_dir="../data", batch_size=16, img_size=64)

imgs, attrs = next(iter(train_loader))
print(imgs.shape, attrs.shape)

imgs_vis = (imgs * 0.5 + 0.5).permute(0,2,3,1).numpy()

plt.figure(figsize=(12,4))
for i in range(8):
    plt.subplot(1,8,i+1)
    plt.imshow(imgs_vis[i])
    plt.axis("off")
    title = [ATTRS[j] for j in range(len(ATTRS)) if attrs[i,j] == 1]
    plt.title(",".join(title), fontsize=8)
plt.show()
```
